Pandas:
-------
Pandas is a Python library used for data manipulation and analysis.

The two most important Pandas structures are:
Pandas

 ├── Series      → 1D(A Series is a one-dimensional labeled data structure.)
 
 └── DataFrame   → 2D(A DataFrame is a two-dimensional labeled data structure)
 

Lesson 1 — Series fundamentals:
-------------------------------

```python
import pandas as pd
marks=pd.Series([80,75,90,85])
print(marks)
```

1.Series vs NumPy Array
----------------------
The major difference is that a Series has an index associated with each value.

2.Custom Index
-------------
You don't have to use 0,1,2,3.

You can create your own labels:

```python
marks=pd.Series(
    [80,75,90],
    index=["Python","SQL","NUMPY"]
)
print(marks)
print(marks.loc["Python"])
```

3.Creating Series from a NumPy Array
------------------------------------

```python
import numpy as np
import pandas as pd

arr=np.array([10,20,30,40])
series=pd.Series(arr)
print(series)
```

4.creating series from a Dictionary
-----------------------------------
dictionary keys become the index and values become the Series values.

```python
marks={
    "python":85,
    "sql":90,
    "NumPy":88
}
series=pd.Series(marks)
print(series)
print(marks["python"]) 
print(series.mean())
```

Accessing Series Values
------------------------
Using Index Position

.loc → label

.iloc → integer position
<------------------------->
Dictionary → marks["Python"]

Series → series.loc["Python"]

```python
import pandas as pd
marks=pd.Series([80,75,90,85])

print(marks.iloc[0])
print(marks.iloc[2])
```

Series Operations
--------------------

```python
marks=pd.Series(
    [80,75,90,85],
    index=["A","B","C","D"]
)
print(marks+5)
print(marks.mean())
```

🎯 Practice

```python
import pandas as pd
marks=pd.Series(
    [85,72,91,68],
    index=["Python","Sql","Numpy","Pandas"]
)
print("marks:")
print(marks)
print("Python:",marks.loc["Python"])
print("First element:",marks.iloc[0])
print("Average_marks:",marks.mean())
print("Highest_marks:",marks.max())
great_75=marks[marks>75]
print("Greater than 75:")
print(great_75)
```

Selecting Multiple Values with .loc
----------------------------------

```python
import pandas as pd 
marks=pd.Series(
    [85,72,91,68,88],
    index=["Python","SQl","NumPy","Pandas","PostgreSQL"]
)
print(marks)
print("------------")
print(marks.loc[["Python","NumPy","PostgreSQL"]])
print("------------")
print(marks.iloc[[1,3]])
```

Slicing with .loc
------------------
iloc → position-based → stop excluded

.loc → label-based → endpoint label included

```python
marks.loc["Python":"NumPy"]
```

```python
marks.iloc[1:3]
```

Boolean Filtering
-----------------
& → AND

| → OR

~ → NOT

IN pandas don't use AND/OR/NOT.use &/|/~

```python
marks[(marks >= 80) & (marks <= 90)]
```

🎯 Your Practice
-----------------

```python
marks = pd.Series(
    [85, 72, 91, 68, 88],
    index=["Python", "SQL", "NumPy", "Pandas", "PostgreSQL"]
)
print("PostgreSQL:",marks.loc["PostgreSQL"])
print("NumPY:",marks.iloc[2])
print("------------------")
print("Select Python and PostgreSQL using .loc:")
print(marks.loc[["Python","PostgreSQL"]])
print("------------------")
print("Select the first three subjects using .iloc.")
print(marks.iloc[[0,1,2]])
print("------------------")
print(marks[marks>80])
subjects=marks[(marks>=70)&(marks<=90)]
print("------------------")
print(subjects)
```

🎯 Your Practice
-----------------

```python
salary = pd.Series(
    [35000, 42000, 55000, 28000, 72000],
    index=["Amit", "Priya", "Rahul", "Neha", "Rohan"]
)
print(salary.dtype)
print(salary.shape)
print(salary.size)
print(salary.index)
print(salary.values)
print("---------------------")
print(salary.head(3))
print("---------------------")
print(salary.tail(2))
print("Average_Salary:",salary.mean())
print("Maximum_Salary:",salary.max())
print("---------------------")
print(salary.describe())
```

| Method       | Purpose                   |
| ------------ | ------------------------- |

| `tail()`     | Last 5 values by default  |

| `max()`      | Maximum                   |
| `min()`      | Minimum                   |
| `sum()`      | Total                     |
| `median()`   | Median                    |
| `std()`      | Standard deviation        |
