rolling()
--------
rolling() calculates a value over a moving window.

pd.date_range() is a Pandas function used to generate a sequence of dates.

```python
import pandas as pd
df=pd.DataFrame({
    "Date":pd.date_range("2026-01-01",periods=6),
    "Sales":[100,120,150,130,170,200]
})
print(df)
```

```python
print(pd.date_range("2026-01-01", periods=6, freq="D"))  # daily
print(pd.date_range("2026-01-01", periods=6, freq="W"))  # weekly
print(pd.date_range("2026-01-01", periods=6, freq="ME")) # month-end
```

```python
df["Rolling_Avg"]=df["Sales"].rolling(3).mean()
print(df)
```

rolling(3) = look at the current value + previous 2 values.

rolling().sum()
---------------

```python
df["Rolling_Sum"]=df["Sales"].rolling(3).sum()
print(df)
```

expanding()
----------
expanding() uses all values from the beginning up to the current row.

```python
df["Cumulative_Avg"]=df["Sales"].expanding().mean()
print(df)
```

ewm() — Exponentially Weighted Moving
-------------------------------------
ewm() gives more weight to recent observations.

```python
df["EWM_Avg"]=df["Sales"].ewm(span=3).mean()
print(df)
```

🧠 The easiest way to remember
| Method        | Meaning                       |
| ------------- | ----------------------------- |
| `rolling()`   | Moving window                 |
| `expanding()` | From beginning → current      |
| `ewm()`       | Recent values get more weight |


Practice 📝
-----------

```python
import pandas as pd
df=pd.DataFrame({
    "Date":pd.date_range("2026-01-01",periods=7),
    "Sales":[100,120,150,130,170,200,180]
})
print(df)
```

```python
df["Rolling_Avg"]=df["Sales"].rolling(3).mean()
df["Rolling_Sum"]=df["Sales"].rolling(3).sum()
df["Expanding_Avg"]=df["Sales"].expanding().mean()
df["Expanding_Sales"]=df["Sales"].expanding().sum()
df["EWM_Avg"]=df["Sales"].ewm(span=3).mean()
print(df)
```
