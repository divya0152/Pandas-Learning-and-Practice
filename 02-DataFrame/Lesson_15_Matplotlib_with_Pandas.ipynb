Lesson 1 — Matplotlib Introduction
------------------------------


Matplotlib
----------
Matplotlib is a Python library used to create visualizations such as:

📈 Line charts,
📊 Bar charts,
📉 Histograms,
🔵 Scatter plots,
🥧 Pie charts,
📦 Box plots,

Think of it like:

Pandas → prepares and analyzes the data

Matplotlib → visualizes the data

```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.DataFrame({
    "Month":["Jan","Feb","Mar","Apr","May"],
    "Sales":[20000,25000,22000,30000,35000]
})
df
```

```python
plt.plot(df["Month"],df["Sales"])
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

Lesson 3: Bar Chart 📊
---------------
A bar chart is mainly used to compare values across different categories.

```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.DataFrame({
    "Department":["IT","HR","Finance","Sales"],
    "Employees":[40,25,30,50]
})
```

```python
plt.bar(df["Department"],df["Employees"])
plt.title("Employees By Department")
plt.xlabel("Department")
plt.ylabel("Number Of Employees")

plt.show()
```

```python
plt.barh(df["Department"],df["Employees"])
plt.title("Employees by Departemnt")
plt.xlabel("Number of Employees")
plt.ylabel("Department")
plt.show()
```

Bar Width
---------

```python
plt.bar(
    df["Department"],
    df["Employees"],
    width=0.5
)
plt.show()
```

Add Values on Bars
-----------------

```python
plt.bar(df["Department"],df["Employees"])

for i, value in enumerate(df["Employees"]):
    plt.text(i,value,str(value),ha="center")
plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.show()
```

Lesson 4:Histogram
---------
A histogram shows how numerical values are distributed across different ranges, called bins.

Histogram = distribution of numerical data

```python
import pandas as pd
import matplotlib.pyplot as plt
Salary = [25000, 28000, 30000, 32000, 35000, 36000,
          40000, 42000, 45000, 48000, 50000, 55000]
plt.hist(Salary)
plt.show()
```

```python
plt.hist(Salary,bins=5)
```

| Histogram                    | Bar Chart                     |
| ---------------------------- | ----------------------------- |
| Numerical data               | Categorical data              |
| Shows distribution           | Shows comparison              |
| Bars usually touch           | Bars usually have gaps        |
| Uses bins/ranges             | Uses categories               |
| Example: salary distribution | Example: salary by department |

Lesson 6 —Pie Charts
-----------
Pie Charts, which are useful for showing part-to-whole proportions.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.DataFrame({
    "Category": ["Rent", "Salary", "Marketing", "Utilities"],
    "Amount": [20000, 50000, 15000, 10000]
})
plt.pie(df["Amount"],labels=df["Category"],autopct="%1.1f%%")
plt.title("Expense Distribution")
plt.show()
```

Lesson 7 — Box Plot
-------------------
A box plot is used to understand the distribution and spread of numerical data and to identify outliers.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.DataFrame({
    "Category": ["Rent", "Salary", "Marketing", "Utilities"],
    "Amount": [20000, 50000, 15000, 10000]
})
```

```python
plt.boxplot(df["Amount"])
plt.title("Salary Distribution")
plt.ylabel("Amount")
plt.show()
```

| Histogram                          | Box Plot                              |
| ---------------------------------- | ------------------------------------- |
| Shows distribution using bins      | Summarizes distribution statistically |
| Shows frequency                    | Shows median, quartiles, IQR          |
| Good for seeing distribution shape | Excellent for detecting outliers      |
| Uses numerical data                | Uses numerical data                   |

Lesson 9 — Figure Size & Subplots
------------------------------
Figure Size
----------

```python
plt.figure(figsize=(5, 2))
#5-->width
#2-->height
plt.plot(df["Month"], df["Sales"], marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

Subplots
--------
Subplots allow you to put multiple charts inside one figure.

Syntax:
-----
plt.subplot(rows, columns, position)

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.plot(df["Month"],df["Sales"])
plt.title("Sales")

plt.subplot(1,2,2)
plt.plot(df["Month"],df["Expenses"])
plt.title("Expenses")

plt.show()
```

plt.subplots()
-------------

```python
fig,axes=plt.subplots(1,2,figsize=(10,4))
axes[0].plot(df["Month"],df["Sales"])
axes[0].set_title("Sales")

axes[1].plot(df["Month"],df["Expenses"])
axes[1].set_title("Expenses")
plt.show()
```

Lesson 10 — Pandas .plot()
-------------------------
Pandas .plot() = a convenient wrapper around Matplotlib

| `kind`      | Chart                |
| ----------- | -------------------- |
| `"line"`    | Line chart           |
| `"bar"`     | Vertical bar chart   |
| `"barh"`    | Horizontal bar chart |
| `"hist"`    | Histogram            |
| `"scatter"` | Scatter plot         |
| `"box"`     | Box plot             |
| `"pie"`     | Pie chart            |


```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.DataFrame({
    "Month":["Jan","Feb","Mar","Apr","May"],
    "Sales":[20000,25000,22000,30000,35000]
})
df.plot(
    x="Month",
    y="Sales",
    kind="line",
    marker="o"
)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
df.plot(
    x="Month",
    y="Sales",
    kind="bar"
)
plt.title("Monthly Sales")
plt.show()
```

Lesson 11 — GroupBy + Visualization 📊
-----------------------------------

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "Sales", "Sales"],
    "Employee": ["A", "B", "C", "D", "E", "F"],
    "Salary": [50000, 60000, 45000, 50000, 40000, 55000]
})

print(df)
```

```python
avg_salary=df.groupby("Department")["Salary"].mean()
print(avg_salary)
```

```python
avg_salary.plot(kind="bar")
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average salary")
plt.show()
```

```python
df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "Sales", "Sales"],
    "Sales": [50000, 60000, 30000, 40000, 70000, 80000]
})
```

```python
total_sales=df.groupby("Department")["Sales"].sum()
print(total_sales)
```

```python
total_sales.plot(kind="bar")
plt.title("Total Sales by Department")
plt.xlabel("Department")
plt.ylabel("Total sales")
plt.show()
```

```python
summary=df.groupby("Department")["Sales"].agg(
    ["sum","mean","count"]
)
print(summary)
summary["mean"].plot(kind="bar")

plt.title("Average Sales by Department")
plt.xlabel("Department")
plt.ylabel("Average Sales")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [20000, 25000, 22000, 30000, 35000, 38000],
    "Expenses": [12000, 14000, 15000, 18000, 20000, 22000]
})

df.plot(
    x="Month",
    y=["Sales", "Expenses"],
    kind="line",
    marker="o"
)

plt.title("Monthly Sales Vs Expenses")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.grid()
plt.legend()
plt.tight_layout()
plt.show()
```

| Question                                           | Suitable chart |
| -------------------------------------------------- | -------------- |
| How does sales change over time?                   | Line           |
| Which department has more employees?               | Bar            |
| What is the salary distribution?                   | Histogram      |
| Are experience and salary related?                 | Scatter        |
| What percentage of expenses goes to each category? | Pie            |
| Are there potential salary outliers?               | Box            |
| Compare many categories horizontally               | Barh           |


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "Finance", "Finance", "Sales", "Sales"],
    "Salary": [60000, 70000, 45000, 50000, 55000, 65000, 50000, 60000]
})
avg_salary=df.groupby("Department")["Salary"].mean()
print(avg_salary)
ax=avg_salary.plot(
    kind="bar",
    figsize=(8,5)
)
for container in ax.containers:
    ax.bar_label(container)
plt.title("Average Salary By Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.tight_layout()
plt.show()
```
