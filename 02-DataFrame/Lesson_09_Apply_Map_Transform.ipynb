Lesson 18 — apply(), map() & transform()
--------------------------------------

map() — change each value
-------------------------

```python
import pandas as pd
df=pd.DataFrame({
    "Name":["Amit","Priya","Rahul"],
    "Salary":[40000,50000,60000]
})
```

```python
df["Salary"]=df["Salary"].map(lambda x:x*1.10)
print(df["Salary"])
```

apply() — custom operation
-------------------------

```python
df.apply(lambda x:x*2)
```

```python
df = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul", "Neha"],
    "Department": ["IT", "HR", "IT", "HR"],
    "Salary": [50000, 40000, 70000, 60000]
})
```

transform() — especially useful with groupby()
---------------------------------------

```python
df["Dept_Avg_Salary"]=df.groupby("Department")["Salary"].transform("mean")
print(df)
```

🧠 Interview memory
-------------------
groupby().mean() → reduces data

transform() → keeps the original number of rows

🎯 Practice
------------

```python
df = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul", "Neha", "Riya"],
    "Department": ["IT", "HR", "IT", "HR", "IT"],
    "Salary": [50000, 40000, 70000, 60000, 55000]
})
```

```python
df["Salary"]=df["Salary"].map(lambda x:x*1.10)
print(df["Salary"])
print("-------------------------")
df["Salary_After_Tax"]=df["Salary"].apply(lambda x:x*0.90)
print(df)
print("-------------------------")
df["Dept_Avg_Salary"]=df.groupby("Department")["Salary"].transform("mean")
#print(df["Dept_Avg_Salary"])
print(df)
```

| Method        | Main purpose                                            |
| ------------- | ------------------------------------------------------- |
| `map()`       | Transform individual values                             |
| `apply()`     | Apply custom function to Series/rows/columns            |
| `transform()` | Group-based calculation while preserving original shape |
