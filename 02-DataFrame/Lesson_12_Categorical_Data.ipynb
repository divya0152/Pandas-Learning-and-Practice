Lesson 21 — Categorical Data in Pandas
-------------------------------------

```python
import pandas as pd
df=pd.DataFrame({
    "Name":["Amit","Priya","Rahul","Neha"],
    "Department":["IT","HR","IT","Finance"]
})
print(df.dtypes)
print("------------------------------------------")
df["Department"]=df["Department"].astype("category")
print(df.dtypes)
```

```python
df["Department"].cat.categories
```

cat is the categorical accessor, similar to .str for strings and .dt for dates.

```python
df["Department"]=df["Department"].cat.add_categories(["Marketing"])
```

```python
print(df["Department"])
```

Rename categories
-----------------

```python
df["Department"]=df["Department"].cat.rename_categories({
    "IT":"Information Technology",
    "HR":"Human Resources",
    "Finance":"Finance"
})
```

Ordered categories
------------------

```python
df["Priority"]=pd.Categorical(
    ["Low","High","Medium","Low"],
    categories=["Low","Medium","High"],
    ordered=True
)
df.sort_values("Priority")
```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul", "Neha", "Riya", "Karan"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Status": ["Active", "Active", "Inactive", "Active", "Inactive", "Active"],
    "Performance": ["High", "Medium", "Low", "High", "Medium", "High"]
})
print(df.dtypes)
```

```python
df["Department"]=df["Department"].astype("category")
print(df)
```

```python
df["Department"].cat.categories
```

```python
df["Status"]=df["Status"].astype("category")
```

```python
df["Performance"]=pd.Categorical(
    df["Performance"],
    categories=["Low","Medium","High"],
    ordered=True
)
print(df)
```

```python
df=df.sort_values("Performance")
print(df)
```

MultiIndex
----------
 A normal DataFrame has one index:
 A Multilndex has multiple levels of index.
 

```python
import pandas as pd

df=pd.DataFrame({
    "Department":["IT","IT","HR","HR"],
    "YEAR":[2024,2025,2024,2025],
    "Salary":[50000,60000,45000,55000]
})
df=df.set_index(["Department","YEAR"])
print(df)
```

Selecting data
--------------

```python
df.loc["IT"]
```

```python
df.loc[("IT",2024)]
```

```python
df=df.reset_index()
print(df)
```

set_index() → columns become index

reset_index() → index becomes columns


MultiIndex from groupby()
-------------------------

```python
result=df.groupby(["Department","YEAR"])["Salary"].mean()
print(df)
```
