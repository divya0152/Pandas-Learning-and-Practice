Employee Information
-------------------

```python
import pandas as pd

employees = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104],
    "Name": ["Amit", "Priya", "Rahul", "Neha"],
    "Department": ["IT", "HR", "IT", "Finance"]
})
```

Salary information
------------------

```python
salary = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104],
    "Salary": [50000, 45000, 70000, 38000]
})
```

```python
result=pd.merge(employees,salary,on="Employee_ID")
print(result)
```

merge() with different key names
----------------------------------
Sometimes the common columns have different names.

```python
employees=pd.DataFrame({
    "Employee_ID":[101,102,103],
    "Name":["Amit","Priya","Rahul"]
})
salary=pd.DataFrame({
    "ID":[101,102,103],
    "Salary":[50000,45000,70000]
})
```

```python
pd.merge(employees,salary,left_on="Employee_ID",right_on="ID")
```

Types of Merge
--------------
The four important types are:
| Merge   | Meaning                          |
| ------- | -------------------------------- |
| `inner` | Only matching records            |
| `left`  | All records from left DataFrame  |
| `right` | All records from right DataFrame |
| `outer` | All records from both DataFrames |


```python
pd.merge(employees,salary,on="Employee_ID",how="inner")
```

concat()
--------
concat() is used to combine DataFrames by stacking them.

merge() → combine based on a common column/key
concat() → stack DataFrames together


Vertical Concatenation
--------------------


```python
import pandas as pd

jan=pd.DataFrame({
    "Employee_ID":[101,102],
    "Name":["Amit","Priya"],
    "Salary":[50000,45000]
})
feb=pd.DataFrame({
    "Employee_ID":[103,104],
    "Name":["Rahul","Neha"],
    "Salary":[70000,38000]
})
result=pd.concat([jan,feb])
print(result)
```

ignore_index=True
-----------------
Usually, when stacking datasets, we want a fresh continuous index

```python
result = pd.concat([jan, feb], ignore_index=True)
print(result)
```

Horizontal Concatenation
-----------------------
concat() can also combine DataFrames side by side.

```python
df1=pd.DataFrame({
    "Name":["Amit","Priya","Rahul"]
})
df2=pd.DataFrame({
    "Salary":[50000,45000,70000]
})
result=pd.concat([df1,df2],axis=1)
print(result)
```

merge()vs concat()
-----------------

| `merge()`                      | `concat()`                   |
| ------------------------------ | ---------------------------- |
| Combines based on key/column   | Stacks/combines DataFrames   |
| Similar to SQL JOIN            | Similar to stacking datasets |
| Uses `on=`                     | Uses `axis=`                 |
| Can use inner/left/right/outer | Can use vertical/horizontal  |

join()
-------
join() combines DataFrames based primarily on their index.

merge() → usually based on columns/keys

join() → usually based on index

```python
df1 = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul"]
}, index=[101, 102, 103])

df2 = pd.DataFrame({
    "Salary": [50000, 45000, 70000]
}, index=[101, 102, 103])
```

```python
result=df1.join(df2)
print(result)
```

| `merge()`                            | `join()`                             |
| ------------------------------------ | ------------------------------------ |
| Usually combines using columns/keys  | Usually combines using index         |
| Similar to SQL JOIN                  | Convenient for index-based combining |
| `on="Employee_ID"`                   | Uses index by default                |
| More flexible for key-based matching | Simpler for index-based matching     |


```python
employees = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul"],
    "Department": ["IT", "HR", "IT"]
}, index=[101, 102, 103])

salary = pd.DataFrame({
    "Salary": [50000, 45000, 70000]
}, index=[101, 102, 103])
```

merge() vs join() vs concat()
--------------------------
| Function   | Main purpose             | Matching based on |
| ---------- | ------------------------ | ----------------- |
| `merge()`  | Combine related datasets | Column/key        |
| `join()`   | Combine related datasets | Index             |
| `concat()` | Stack/combine DataFrames | Rows or columns   |
