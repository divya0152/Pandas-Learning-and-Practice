Lesson 19 — Datetime & Time Series
-----------------------------------

```python
import pandas as pd
df=pd.DataFrame({
    "Name":["Amit","Priya","Rahul"],
    "Joining_Date":["2024-01-15","2023-06-20","2025-03-10"]
})
df["Joining_Date"]=pd.to_datetime(df["Joining_Date"])
print(df)
print(df.dtypes)
```

Extract Year, Month & Day
--------------------------

```python
df["Year"]=df["Joining_Date"].dt.year
print(df)
```

```python
df["Day_Name"] = df["Joining_Date"].dt.day_name()
print(df)
```

Date Filtering
--------------

```python
df=pd.DataFrame({
    "Name":["Amit","Priya","Rahul","Neha"],
    "Joining_Date":pd.to_datetime([
        "2022-01-10",
        "2023-05-15",
        "2024-07-20",
        "2025-03-12"
    ])
})
```

```python
df[df["Joining_Date"]>"2023-01-01"]
```

Sorting by Date
---------------

```python
df=df.sort_values("Joining_Date")
print(df)
```

```python
df=df.sort_values("Joining_Date",ascending=False)
print(df)
```

Date Difference
--------------

```python
df = pd.DataFrame({
    "Name": ["Amit", "Priya"],
    "Joining_Date": pd.to_datetime(["2022-01-10", "2024-05-15"])
})

today = pd.Timestamp("2026-09-10")
```

```python
df["Days_At_Company"]=today-df["Joining_Date"]
print(df)
```

```python
df["Days_At_Company"].dt.days
```

Datetime as an Index
--------------------

```python
import pandas as pd
df=pd.DataFrame({
    "Date":pd.to_datetime([
        "2026-01-01",
        "2026-01-02",
        "2026-01-03"
    ]),
    "Sales":[100,120,150]
})
df=df.set_index("Date")
print(df)
```

Select Data by Date
-------------------

```python
df.loc["2026-01-02"]
```

```python
df.loc["2026-01-01":"2026-01-02"]
```

----------------------------

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul", "Neha", "Riya", "Karan"],
    "Department": ["IT", "HR", "IT", "HR", "IT", "Finance"],
    "Age": [24, 21, 27, 23, 25, 29],
    "Salary": [50000, 40000, 70000, 60000, 55000, 45000]
})
```

```python
print(df.groupby("Department")["Salary"].mean())
print("-----------------------------------------")
print(df.groupby("Department")["Salary"].agg(["min","max","mean"]))
print("-----------------------------------------")
df["Dept_Avg_Salary"]=df.groupby("Department")["Salary"].transform("mean")
print(df)
print("-----------------------------------------")
df["Salary_After_Tax"]=df["Salary"].apply(lambda x:x*0.90)
print(df)
print("-----------------------------------------")
df["Department_Code"]=df["Department"].map({
    "IT":1,
    "HR":2,
    "Finance":3
})
print(df)
```

```python
monthly = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul"],
    "Jan": [50000, 40000, 70000],
    "Feb": [52000, 42000, 72000],
    "Mar": [55000, 45000, 75000]
})
result=monthly.melt(
    id_vars=["Name"],
    var_name="Month",
    value_name="Salary"
)
print(result)
    
```

```python
df=pd.DataFrame({
    "Name":["Amit","Priya","Rahul","Neha","Riya"],
    "Joining_Date":[
        "2022-01-15",
        "2023-06-20",
        "2024-03-10",
        "2025-08-05",
        "2023-12-25"
    ],
    "Salary":[40000,50000,60000,45000,55000]
})
print(df)
```

```python
df["Joining_Date"]=pd.to_datetime(df["Joining_Date"])
df["Year"]=df["Joining_Date"].dt.year
df["Month"]=df["Joining_Date"].dt.month
df["Day"]=df["Joining_Date"].dt.day
df["Month_Name"]=df["Joining_Date"].dt.month_name()
print(df)
print("----------------------------------")
print(df[df["Joining_Date"]<"2024-01-01"])
print("----------------------------------")
print(df.sort_values("Joining_Date",ascending=False))
print("----------------------------------")
df=df.set_index("Joining_Date")
print(df)
```
