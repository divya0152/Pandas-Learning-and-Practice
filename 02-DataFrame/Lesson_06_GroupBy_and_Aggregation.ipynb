Lesson 15 — groupby() & Aggregation
---------------------------------
groupby() is used to divide data into groups and then perform calculations on each group.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul", "Neha", "Riya"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [50000, 45000, 70000, 38000, 60000]
})

print(df)
```

```python
df.groupby("Department")["Salary"].mean()
```

```python
df.groupby("Department")["Salary"].sum()
```

```python
df.groupby("Department")["Name"].count()
```

```python
df.groupby("Department")["Salary"].max()
```

```python
df.groupby("Department")["Salary"].min()
```

```python
df.groupby("Department")["Salary"].agg(["mean","max","min"])
```

Grouping by multiple columns.
------------------------------

```python
import pandas as pd

df = pd.DataFrame({
    "Name": [
        "Amit", "Priya", "Rahul", "Neha", "Riya",
        "Karan", "Sneha", "Vikas", "Pooja", "Arjun"
    ],
    "Department": [
        "IT", "HR", "IT", "Finance", "HR",
        "IT", "Finance", "HR", "IT", "Finance"
    ],
    "Age": [24, 21, 27, 23, 25, 29, 26, 22, 31, 28],
    "Salary": [
        50000, 45000, 70000, 38000, 60000,
        75000, 48000, 52000, 65000, 55000
    ]
})
```

```python
print(df.groupby("Department")["Salary"].mean())
print("----------------------------------------")
print(df.groupby("Department")["Salary"].sum())
print("----------------------------------------")
print(df.groupby("Department")["Name"].count())
print("----------------------------------------")
print(df.groupby("Department")["Salary"].max())
print("----------------------------------------")
print(df.groupby("Department")["Salary"].min())
print("----------------------------------------")
print(df.groupby("Department")["Salary"].agg(["mean","max","min"]))
print("----------------------------------------")
print(df.groupby(["Department","Age"])["Salary"].sum())
print("----------------------------------------")
print(df.groupby(["Department","Age"])["Salary"].mean())
```

reset_index():
-------------
It converts the DataFrame/Series index back into a normal column and creates a new default index (0, 1, 2, ...).


```python
df.groupby("Department")["Salary"].mean().reset_index()
```

Lesson 15 — groupby() & Aggregation: agg() with Multiple Columns
----------------------------------------------------------------

```python
df.groupby("Department").agg({
    "Salary":"sum",
    "Age":"mean"
})
```

```python
df.groupby("Department").agg({
    "Salary":["min","max","mean"],
    "Age":["min","max","mean"]
})
```

groupby() is used to divide the DataFrame into groups based on one or more columns, while agg() is used to apply one or more aggregation functions to those groups.
