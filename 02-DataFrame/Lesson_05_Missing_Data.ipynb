📓 Lesson 10 — Handling Missing Data
------------------------------------

```python
import numpy as np
import pandas as pd

df=pd.DataFrame({
    "Name":["Amit","Priya","Rahul","Neha"],
    "Age":[21,np.nan,23,24],
    "Salary":[40000,45000,np.nan,"IT"]
})
print(df)
```

Detect Missing Values — isna()
-------------------------------
df.isna()

This returns True where a value is missing and False where it isn't.

```python
df.isna()
```

Count Missing Values
--------------------


```python
df.isna().sum()
```

Find Rows With Missing Values
----------------------------

```python
df[df["Age"].isna()]
```

```python
df[df["Salary"].isna()]
```

notna() — Find Non-Missing Values
---------------------------------
notna() is the opposite of isna()

isna()   → Is it missing?

notna()  → Is it NOT missing?

```python
df[df["Age"].notna()]
```

-------------

dropna()
--------
dropna() is used when you want to remove rows or columns containing missing values.

By default, Pandas removes any row that contains at least one missing value

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul", "Neha"],
    "Age": [21, np.nan, 23, 24],
    "Salary": [40000, 45000, np.nan, 38000],
    "Department": ["IT", "HR", np.nan, "IT"]
})

print(df)
```

```python
df.dropna()
```

axis tells Pandas whether to remove rows or columns

how controls when a row/column should be removed.


| Code                           | Meaning                               |
| ------------------------------ | ------------------------------------- |
| `df.dropna()`                  | Remove rows containing missing values |
| `df.dropna(axis=0)`            | Remove rows                           |
| `df.dropna(axis=1)`            | Remove columns                        |
| `df.dropna(how="any")`         | Remove if any value is missing        |
| `df.dropna(how="all")`         | Remove only if all values are missing |
| `df.dropna(subset=["Salary"])` | Remove rows where Salary is missing   |


```python
print(df.dropna(how="any"))
print("-----------------------------")
print(df.dropna(subset="Salary"))
```

fillna()
---------

```python
import pandas as pd
import numpy as np
df=pd.DataFrame({
    "Name":["Amit","Priya","Rahul","Neha"],
    "Age":[21,np.nan,23,24],
    "Salary":[40000,45000,np.nan,38000]
})
#df["Age"]=df["Age"].fillna(25)
print(df)
```

Fill missing values in the whole DataFrame
------------------------------------------

| Data                           | Common filling method |
| ------------------------------ | --------------------- |
| Numerical, relatively balanced | Mean                  |
| Numerical, outliers/skewed     | Median                |
| Categorical                    | Mode                  |


```python
df=df.fillna(0)
print(df)
```

```python
df["Salary"].median()
df["Salary"]=df["Salary"].fillna(df["Salary"].median())
print(df)
```

```python
df["Age"].mean()
df["Age"]=df["Age"].fillna(df["Age"].mean())
print(df)
```

```python
df["Age"]=df["Age"].astype(int)
df["Salary"]=df["Salary"].astype(int)
print(df.dtypes)
```

Lesson 11 — Removing Duplicates
-------------------------------
A duplicate row is a row that appears more than once in your DataFrame.

```python
df=pd.DataFrame({
    "Name":["Amit","Priya","Rahul","Amit"],
    "Age":[21,22,23,21],
    "Salary":[40000,45000,50000,40000]
})
df.duplicated()
df.duplicated().sum()
print(df)
```

```python
df=df.drop_duplicates()
print(df)
```

📝 Practice
-------------

```python
df = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 101, 104, 102],
    "Name": ["Amit", "Priya", "Rahul", "Amit", "Neha", "Priya"],
    "Salary": [40000, 45000, 50000, 40000, 38000, 45000]
})
print("Find which rows are duplicates")
print(df.duplicated())
print("Count the number of duplicate rows:",df.duplicated().sum())
df.drop_duplicates()
df.duplicated(subset=["Employee_ID"])
```

duplicated() is used to identify duplicate rows and returns a Boolean Series, while drop_duplicates() removes duplicate rows from the DataFrame
| Topic                     | What to remember                                      | Syntax / Example                                      |
| ------------------------- | ----------------------------------------------------- | ----------------------------------------------------- |
| **Missing data**          | Missing values are commonly represented as `NaN`      | `np.nan`                                              |
| **Detect missing values** | `isna()` returns `True` for missing values            | `df.isna()`                                           |
| **Count missing values**  | Count missing values in each column                   | `df.isna().sum()`                                     |
| **`isnull()`**            | Same functionality as `isna()`                        | `df.isnull()`                                         |
| **Find missing rows**     | Filter rows where a column is missing                 | `df[df["Age"].isna()]`                                |
| **Find non-missing rows** | Use `notna()`                                         | `df[df["Age"].notna()]`                               |
| **`dropna()`**            | Removes rows/columns containing missing values        | `df.dropna()`                                         |
| **Remove rows**           | `axis=0` means rows                                   | `df.dropna(axis=0)`                                   |
| **Remove columns**        | `axis=1` means columns                                | `df.dropna(axis=1)`                                   |
| **`how="any"`**           | Remove if **at least one** value is missing           | `df.dropna(how="any")`                                |
| **`how="all"`**           | Remove only if **all** values are missing             | `df.dropna(how="all")`                                |
| **`subset`**              | Check missing values only in specific column(s)       | `df.dropna(subset=["Salary"])`                        |
| **`fillna()`**            | Replaces missing values instead of deleting rows      | `df["Age"].fillna(25)`                                |
| **Fill with fixed value** | Replace missing values with a chosen value            | `df["Age"] = df["Age"].fillna(25)`                    |
| **Fill with mean**        | Useful for numerical data when appropriate            | `df["Age"].fillna(df["Age"].mean())`                  |
| **Fill with median**      | Useful when data has outliers/skew                    | `df["Salary"].fillna(df["Salary"].median())`          |
| **Fill with mode**        | Common for categorical data                           | `df["Department"].fillna(df["Department"].mode()[0])` |
| **Mean vs median**        | Mean is affected by outliers; median is less affected | Salary → often median                                 |
| **Outlier**               | An unusually extreme value                            | `500000` among salaries around `30–40k`               |
| **Skewed data**           | Distribution is uneven / has a long tail              | High-salary values can create right skew              |

Lesson 14 — Sorting Data in Pandas
----------------------------------
sort_values() → sort based on column values

sort_index() → sort based on row/column index

```python
import pandas as pd
df=pd.DataFrame({
    "Name":["Amit","Priya","Rahul","Neha","Riya"],
    "Age":[24,21,27,23,25],
    "Salary":[50000,45000,70000,38000,60000]
})
df.sort_values("Salary")
```

```python
#By default, sorting is ascending.
df.sort_values("Salary",ascending=False)
```

Sort by multiple columns
-----------------------

```python
df.sort_values(["Name","Salary"])
```

```python
df.sort_values(["Name","Salary"],ascending=[True,False])
```

sort_index()
-------------
sort_index() sorts the DataFrame based on its index.

```python
df=pd.DataFrame({
    "Name":["Amit","Priya","Rahul"],
    "Salary":[50000,45000,70000]
},index=[3,1,2])
print(df)
```

```python
df.sort_index()
```

sort_values() vs sort_index()
------------------------------

| Method          | Sorts by          |
| --------------- | ----------------- |
| `sort_values()` | **Column values** |
| `sort_index()`  | **Index labels**  |


```python
df = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul", "Neha", "Riya"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Age": [24, 21, 27, 23, 25],
    "Salary": [50000, 45000, 70000, 38000, 60000]
})
print(df.sort_values(["Salary"]))
print("________________________")
print(df.sort_values(["Salary"],ascending=False))
print("-------------------------")
print(df.sort_values(["Age"]))
print("--------------------------")
print(df.sort_values(["Department","Salary"],ascending=[True,False]))
df.index = [5, 3, 1, 4, 2]
print(df.sort_index())
```
