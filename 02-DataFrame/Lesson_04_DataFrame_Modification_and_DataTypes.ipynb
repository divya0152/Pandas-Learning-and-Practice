Lesson 8 — Adding, Updating & Deleting DataFrame Data
----------------------------------------------------

```python
import pandas as pd

data = {
    "Name": ["Amit", "Priya", "Rahul", "Neha", "Rohan"],
    "Python": [85, 78, 92, 68, 88],
    "SQL": [90, 82, 95, 70, 91],
    "Salary": [40000, 45000, 60000, 35000, 55000]
}

df = pd.DataFrame(data)
```

Adding a New Column
-------------------

```python
df["Experience"]=[2,3,5,1,4]
print(df)
```

Creating a Calculated Column
--------------------------

```python
df["Bonus"]=5000
df["Total_Salary"]=df["Salary"]+df["Bonus"]
print(df)
```

Updating a Column
-----------------

```python
df["Bonus"]=2000
df["Salary"] = df["Salary"] + 3000
```

Updating Specific Rows with .loc()
----------------------------------
df.loc[condition, "column"] = new_value

```python
df.loc[df["Name"]=="Neha","Salary"]=38000
```

```python
print(df)
```

Updating Multiple Rows
----------------------

```python
df.loc[df["Salary"]<40000,"Salary"]=40000
print(df)
```

Removing a Column
-----------------

```python
df=df.drop(columns=["Bonus"])
```

Removing Multiple Columns
-------------------------


```python
df=df.drop(columns=["Total_Salary", "Experience"])
print(df)
```

```python
print(df)
```

Removing Rows
-------------

```python
df = df.drop(index=3)
```

```python
df=df.drop(index=[1,4])
print(df)
```

rename()
--------

```python
df=df.rename(columns={
    "Python":"Python_Marks",
    "SQL":"SQL_Marks"
})
print(df)
```

```python
df=df.assign(
    Bonus=5000
)
```

```python
df = df.assign(
    Total_Salary=df["Salary"] + 5000
)
```

```python
print(df)
```

🧠 Quick Summary
----------------

| Operation               | Pandas                                     |
| ----------------------- | ------------------------------------------ |
| Add column              | `df["Bonus"] = 5000`                       |
| Calculated column       | `df["Total"] = df["Salary"] + df["Bonus"]` |
| Update specific row     | `df.loc[condition, "Salary"] = value`      |
| Remove column           | `df.drop(columns=["Bonus"])`               |
| Remove row              | `df.drop(index=3)`                         |
| Rename column           | `df.rename(columns={...})`                 |
| Create using `assign()` | `df.assign(Bonus=5000)`                    |

Lesson 9 — Data Types & astype()
--------------------------------

A data type tells Pandas what kind of data is stored in a column.
| Pandas dtype | Meaning                                                | Example        |
| ------------ | ------------------------------------------------------ | -------------- |
| `int64`      | Integer                                                | `25`           |
| `float64`    | Decimal number                                         | `25.5`         |
| `object`     | General/mixed data, commonly older text representation | `"Amit"`       |
| `string`     | Text                                                   | `"Amit"`       |
| `bool`       | True/False                                             | `True`         |
| `datetime64` | Date/time                                              | `2026-09-07`   |
| `category`   | Repeated categorical values                            | `"IT"`, `"HR"` |


```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul"],
    "Age": [21, 22, 23],
    "Salary": [40000.5, 45000.0, 60000.5],
    "Active": [True, False, True]
})

print(df.dtypes)
```

astype() — Changing Data Type
-----------------------------

The basic syntax is:

df["column"] = df["column"].astype(new_type)

```python
df["Age"].dtype
```

```python
df["Age"]=df["Age"].astype(int)
df["Age"].dtype
```

```python
df["Salary"]=df["Salary"].astype(int)
df["Salary"].dtype
```

```python
df["Name"]=df["Name"].astype("string")
df["Name"].dtype
```
