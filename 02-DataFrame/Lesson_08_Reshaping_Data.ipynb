```python
import pandas as pd
df = pd.DataFrame({
    "Name": ["Amit", "Priya", "Rahul"],
    "Jan": [100, 200, 150],
    "Feb": [120, 220, 170],
    "Mar": [130, 210, 180]
})
print(df)
```

melt()
-------
converts Wide-->Long format

Wide → more columns

Long → more rows

```python
result=df.melt(
    id_vars=["Name"],
    var_name="Month",
    value_name="Salary"
)
print(result)
```

```python
wide_df = result.pivot(
    index="Name",
    columns="Month",
    values="Salary"
)

print(wide_df)
```

🧠 Easy memory
--------------
melt()  → Wide → Long

pivot() → Long → Wide

```python
import pandas as pd

df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "IT"],
    "Month": ["Jan", "Jan", "Jan", "Feb", "Feb"],
    "Salary": [50000, 60000, 45000, 48000, 70000]
})
```

```python
result=pd.pivot_table(
    df,
    index="Department",
    columns="Month",
    values="Salary",
    aggfunc="mean"
)
print(result)
```

🧠 pivot() vs pivot_table()
----------------------------
| `pivot()`                            | `pivot_table()`       |
| ------------------------------------ | --------------------- |
| Reshapes data                        | Reshapes + aggregates |
| Cannot handle duplicate combinations | Can handle duplicates |
| No aggregation function              | Supports `aggfunc`    |
| Simpler                              | More powerful         |
