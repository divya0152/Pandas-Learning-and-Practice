Choosing Efficient Data Types
---------------------------
Use the smallest appropriate dtype for your data to reduce memory usage and potentially improve performance.

| Data                 | Possible dtype       |
| -------------------- | -------------------- |
| Small whole numbers  | `int8`, `int16`      |
| Larger whole numbers | `int32`, `int64`     |
| Decimal values       | `float32`, `float64` |
| Repeated categories  | `category`           |
| Text                 | `string`             |


```python
import pandas as pd
df = pd.DataFrame({
    "Age": [21, 24, 27, 30, 35],
    "Experience": [1, 3, 5, 7, 10],
    "Salary": [40000, 50000, 60000, 75000, 90000]
})
```

```python
print(df.dtypes)
print(df.memory_usage(deep=True))
```

```python
df["Age"] = df["Age"].astype("int8")
df["Experience"] = df["Experience"].astype("int8")
df["Salary"] = df["Salary"].astype("int32")
print(df.dtypes)
```

| dtype   | Storage |            Approximate range |
| ------- | ------: | ---------------------------: |
| `int8`  |  1 byte |                  -128 to 127 |
| `int16` | 2 bytes |            -32,768 to 32,767 |
| `int32` | 4 bytes | ~-2.1 billion to 2.1 billion |
| `int64` | 8 bytes |             Very large range |

eval()
-----
eval() allows you to perform calculations involving DataFrame columns using an expression written as a string.

```python
df["Total"]=df["Salary"]+df["Bonus"]
df.eval("Total=Salary+Bonus",inplace=True)
print(df)
```

```python
import pandas as pd
df=pd.DataFrame({
    "Name":["Amit","Priya","Rahu"],
    "Salary":[40000,50000,70000],
    "Bonus":[4000,5000,7000]
})
df.eval("Total_Salary=Salary+Bonus",inplace=True)
df.eval("Net_Salary=Total_Salary-2000",inplace=True)
print(df)
```

Benchmarking
------------
Benchmarking means measuring how long code takes to execute.

```python
import timeit
time=timeit.timeit(
    "sum(range(1000))",
    number=1000
)
print(time)
```

```python
import pandas as pd
import numpy as np
import timeit

df=pd.DataFrame({
    "Salary":np.random.randint(30000,100000,100000)
})

vectorized_time=timeit.timeit(
    "df['Salary']*1.10",
    globals=globals(),
    number=100
)
print(vectorized_time)
```

```python
apply_time=timeit.timeit(
    "df['Salary'].apply(lambda x:x*1.10)",
    globals=globals(),
    number=100
)
print(apply_time)
```

Downcasting
-----------
Downcasting means converting a numeric column from a larger dtype to a smaller dtype when the values safely fit.

Downcasting = same type, smaller size

int64 → int32 → int16 → int8
float64 → float32

📝 Practice
------------

```python
import pandas as pd
df = pd.DataFrame({
    "Age": [21, 24, 27, 30, 35],
    "Experience": [1.5, 3.0, 5.5, 7.0, 10.0],
    "Salary": [40000, 50000, 60000, 75000, 90000]
})
```

```python
print(df.dtypes)
```

```python
df["Age"]=pd.to_numeric(df["Age"],downcast="integer")
df["Experience"]=pd.to_numeric(df["Experience"],downcast="float")
```

```python
print(df.dtypes)
```

🧠 Complete memory optimization formula
-----------------------
Load less → Store efficiently → Process less → Process in batches

In Pandas:

Load less → Store efficiently → Filter early → Vectorize → Avoid unnecessary copies → Process in chunks → Measure performance
