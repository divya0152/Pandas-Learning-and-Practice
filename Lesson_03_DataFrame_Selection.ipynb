Lesson 3 — DataFrame Data Selection
-----------------------------------

```python
import pandas as pd

data = {
    "Name": ["Amit", "Priya", "Rahul", "Neha", "Rohan"],
    "Python": [85, 78, 92, 68, 88],
    "SQL": [90, 82, 95, 70, 91],
    "Salary": [40000, 45000, 60000, 35000, 55000]
}

df = pd.DataFrame(data)
```

```python
print("Selecting One Column")
print(df["Python"])
print("Selecting Multiple Columns")
print(df[["Name","Python","Salary"]])
```

```python
print("Selects the row whose index label is 0:")
df.loc[0]
```

```python
print("Selecting Specific Rows and Columns:")
df.loc[0,"Name"]
```

```python
print("Give the first row:")
df.iloc[0]
```

```python
print("First row, second column.")
df.iloc[0,1]
```

```python
print("Give me the names and salaries of Rahul and Rohan.")
df.loc[[2,4],["Name","Salary"]]
```

🎯 Interview Question
----------------------
Why do we use & instead of and when filtering a Pandas DataFrame?

"& is used in Pandas for element-wise AND operations between Series conditions. The Python and operator expects a single Boolean value, whereas a Pandas condition produces multiple Boolean values—one for each row."

Lesson 4-isin() in Pandas
--------------------------
isin(), which is useful when you want to filter rows matching multiple specific values.

```python
print("--Without isin()---")
df[
    (df["Name"] == "Amit") |
    (df["Name"] == "Rahul") |
    (df["Name"] == "Rohan")
]
```

```python

print("--with isin()--")
df[df["Name"].isin(["Amit","Rahul"])]
```

```python
print("---Find employees whose names are:(Amit, Rahul, Rohan)Using .isin()--")
print(df[df["Name"].isin(["Amit","Rahul","Rohan"])])
print("____________")
print("---Find employees whose names are:(Priya or Neha)Using .isin()--")
print(df[df["Name"].isin(["Priya"])|df["Name"].isin(["Neha"])])
print("____________")
print(df[~df["Name"].isin(["Amit","Rahul"])])
print("____________")
print(df[df["Python"].isin([78,88,92])])
```

Lesson5-between() for Range Filtering
--------------------------------
between() checks whether each value in a Pandas Series falls between a lower limit and upper limit.

== → one exact value
    
.isin() → multiple specific values

between() → values within a range

| Code                                   | Meaning     |
| -------------------------------------- | ----------- |
| `between(40, 50)`                      | 40 ≤ x ≤ 50 |
| `between(40, 50, inclusive="both")`    | 40 ≤ x ≤ 50 |
| `between(40, 50, inclusive="neither")` | 40 < x < 50 |
| `between(40, 50, inclusive="left")`    | 40 ≤ x < 50 |
| `between(40, 50, inclusive="right")`   | 40 < x ≤ 50 |


```python
print("Find employees whose Salary is between ₹40,000 and ₹55,000")
print(df[df["Salary"].between(40000,55000)])
print("-------------------------------------")
print("Find employees whose Python marks are between 80 and 90")
print(df[df["Python"].between(80,90)])
print("-------------------------------------")
print("Find employees whose SQL marks are between 85 and 95")
print(df[df["SQL"].between(85,95)])
print("-------------------------------------")
print("Find employees whose salary is between ₹40,000 and ₹60,000, but exclude both boundaries.")
print(df[df["Salary"].between(40000,60000, inclusive="neither")])
```

Lesson 7 — query() for DataFrame Filtering
------------------------------------------
query() lets you filter a DataFrame using a readable condition written as a string.

| Method            | Example                                 | Best use            |
| ----------------- | --------------------------------------- | ------------------- |
| Boolean filtering | `df[df["Salary"] > 50000]`              | General filtering   |
| `.isin()`         | `df[df["Name"].isin([...])]`            | Specific values     |
| `.between()`      | `df[df["Salary"].between(40000,55000)]` | Range               |
| `query()`         | `df.query("Salary > 50000")`            | Readable conditions |


Practice 
-----------

```python
print("Find employees whose Python marks are greater than 80 using query():")
print(df.query("Python>80"))
print("_______________________________")
print("find employees whose Salary is greater than 50,000 AND SQL marks are greater than 90")
print(df.query("Salary>50000 and SQL>90"))
print("_______________________________")
print("Find employees whose Python marks are greater than 90 OR Salary is greater than 50,000")
print(df.query("Python>90 or Salary>50000"))
print("_______________________________")
print("Find employees whose:Python marks are at least 85 AND salary is at least ₹50,000.Using query() only")
print(df.query("Python>=85 and Salary >=50000"))
print("_______________________________")
```
